TRANSFORM

```python
df_clean = df.dropna(subset=["Harga"])
print(f"Baris dibuang: {len(df) - len(df_clean)}")
```

```python
df_clean.describe()
```

```python
df_clean = df_clean.copy()
df_clean["Harga"] = df_clean["Harga"].str.replace("Rp", "", regex=False)
df_clean["Harga"] = df_clean["Harga"].str.replace(",", "", regex=False)
df_clean["Harga"] = pd.to_numeric(df_clean["Harga"])

df_clean[["Nama Provinsi", "Komoditas", "Harga"]].head(5)
```

```python
bulan_map = {
    "Januari": 1, "Februari": 2, "Maret": 3, "April": 4,
    "Mei": 5, "Juni": 6, "Juli": 7, "Agustus": 8,
    "September": 9, "Oktober": 10, "November": 11, "Desember": 12
}

df_clean["Bulan_num"] = df_clean["Bulan"].map(bulan_map)

df_clean["Tanggal"] = pd.to_datetime(
    df_clean[["Tahun", "Bulan_num"]].rename(columns={"Tahun": "year", "Bulan_num": "month"}).assign(day=1)
)

df_clean[["Nama Provinsi", "Bulan", "Tahun", "Tanggal", "Harga"]].head(5)
```

```python
df_clean.head(5)
```

```python
# Buang kolom yang sudah tidak diperlukan
df_final = df_clean.drop(columns=["Bulan", "Tahun", "Bulan_num"])

df_final.head(5)
```

LOAD

```python
import sqlite3

# Buat koneksi ke database (otomatis buat file .db kalau belum ada)
conn = sqlite3.connect("../output/harga_pangan.db")

# Load dataframe ke database
df_final.to_sql("harga_pangan", conn, if_exists="replace", index=False)

print(f"Berhasil load {len(df_final)} baris ke database!")

conn.close()
```

VERIFY

```python
conn = sqlite3.connect("../output/harga_pangan.db")

result = pd.read_sql("""
    SELECT Komoditas, ROUND(AVG(Harga), 0) as Rata_Rata_Harga
    FROM harga_pangan
    GROUP BY Komoditas
    ORDER BY Rata_Rata_Harga DESC
""", conn)

conn.close()

result
```
